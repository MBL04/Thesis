```python
import pandas as pd
import numpy as np
```

```python
path = r"C:\Users\micha\OneDrive\Documents\PRSGWAS_tsv\GCST90104540_buildGRCh37EU.tsv\Meta_IS_BothMandF_GIGA_EA_allfiles_metaAnalyseSEbased_wogc_2021June11_NeNs3rdDp1e6.gwascatalog.tsv_2"
gwas_EUdata = pd.read_csv(path, sep='\t')
display(gwas_EUdata)
```

```python
rename_dict = {
    'chromosome': 'CHR',
    'base_pair_location': 'BP',
    'effect_allele': 'A1',      # A1 is the target/risk allele
    'other_allele': 'A2',       # A2 is the baseline allele
    'beta': 'BETA',
    'p_value': 'P',
    'effect_allele_frequency': 'EAF',
    'standard_error': 'SE',
    'odds_ratio': 'OR'
}
```

```python
eurename = gwas_EUdata.rename(columns=rename_dict)
```

```python
eurename
```

```python
columns_to_keep = ['CHR', 'BP', 'A1', 'A2', 'BETA', 'P', 'EAF', 'SE', 'OR'] 
eu = eurename[columns_to_keep]
eu
```

```python
missing = eu.isna().sum()
print(missing)
```

```python
eu['MAF'] = np.where(
    eu['EAF'] > 0.5,     # The IF condition because Minor Alleles need to be less than 0.5, if effect allele is less than 0.5, calculation does not apply and MAF = EAF
    1 - eu['EAF'],       # What to do if TRUE
    eu['EAF']            # What to do if FALSE
)
```

```python
eu
```

```python
# Find the lowest and highest EAF in your data, for sanity check if calculations were wrong
min_eaf = eu['EAF'].min() 
max_eaf = eu['EAF'].max()

print(f"The lowest EAF is: {min_eaf}")
print(f"The highest EAF is: {max_eaf}")
```

```python
eu = eu[eu['MAF'] >= 0.01] # Filtering SNPs with MAF less than 1%, theory is this may be only noise
```

```python
# Check for duplicate SNPs
print(eu.duplicated(subset=['CHR', 'BP']).sum())
```

```python
min_p = eu['P'].min() 
max_p = eu['P'].max()

print(f"The lowest p is: {min_p}")
print(f"The highest p is: {max_p}")
```

```python
eu = eu[(eu['P'] > 0) & (eu['P'] <= 1)]
```
