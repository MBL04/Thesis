```python
import pandas as pd
import numpy as np
```

```python
path = r"C:\Users\micha\OneDrive\Documents\PRSGWAS_tsv\GCST90104545_buildGRCh37EA.tsv\Meta_IS_BothMandF_GIGA_ASN_allfiles_metaAnalyseSEbased_wogc_2021June11_NeNs3rdDp1e6.gwascatalog.tsv_2"
gwas_EAdata = pd.read_csv(path, sep='\t')
display(gwas_EAdata)
```

```python
rename_dict = {
    'chromosome': 'CHR',
    'base_pair_location': 'BP',
    'effect_allele': 'A1',      # A1 is the target/risk allele
    'other_allele': 'A2',       # A2 is the baseline allele
    'beta': 'BETA',
    'p_value': 'P',
    'effect_allele_frequency': 'EAF',
    'standard_error': 'SE',
    'odds_ratio': 'OR'
}
```

```python
earename = gwas_EAdata.rename(columns=rename_dict)
```

```python
earename
```

```python
columns_to_keep = ['CHR', 'BP', 'A1', 'A2', 'BETA', 'P', 'EAF', 'SE', 'OR'] 
ea = earename[columns_to_keep]
ea
```

```python
missing = ea.isna().sum()
print(missing)
```

```python
ea['MAF'] = np.where(
    ea['EAF'] > 0.5,     # The IF condition because Minor Alleles need to be less than 0.5, if effect allele is less than 0.5, calculation does not apply and MAF = EAF
    1 - ea['EAF'],       # What to do if TRUE
    ea['EAF']            # What to do if FALSE
)
```

```python
ea
```

```python
# Find the lowest and highest EAF in your data, for sanity check if calculations were wrong
min_eaf = ea['EAF'].min() 
max_eaf = ea['EAF'].max()

print(f"The lowest EAF is: {min_eaf}")
print(f"The highest EAF is: {max_eaf}")
```

```python
ea = ea[ea['MAF'] >= 0.01] # Filtering SNPs with MAF less than 1%, theory is this may be only noise
```

```python
ea
```

```python
# Check for duplicate SNPs
print(ea.duplicated(subset=['CHR', 'BP']).sum())
```

```python
ea = ea[(ea['P'] > 0) & (ea['P'] <= 1)]
```
