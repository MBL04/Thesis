```python
import pandas as pd
import numpy as np
```

```python
path = r"C:\Users\micha\OneDrive\Documents\PRSGWAS_tsv\GCST90104560_buildGRCh37SA.tsv\Meta_IS_BothMandF_GIGA_SAS_allfiles_metaAnalyseSEbased_wogc_2021June11_NeNs3rdDp1e6.gwascatalog.tsv_2"
gwas_SAdata = pd.read_csv(path, sep='\t')
display(gwas_SAdata)
```

```python
rename_dict = {
    'chromosome': 'CHR',
    'base_pair_location': 'BP',
    'effect_allele': 'A1',      # A1 is the target/risk allele
    'other_allele': 'A2',       # A2 is the baseline allele
    'beta': 'BETA',
    'p_value': 'P',
    'effect_allele_frequency': 'EAF',
    'standard_error': 'SE',
    'odds_ratio': 'OR'
}
```

```python
sarename = gwas_SAdata.rename(columns=rename_dict)
```

```python
sarename
```

```python
columns_to_keep = ['CHR', 'BP', 'A1', 'A2', 'BETA', 'P', 'EAF', 'SE', 'OR'] 
sa = sarename[columns_to_keep]
sa
```

```python
missing = sa.isna().sum()
print(missing)
```

```python
sa['MAF'] = np.where(
    sa['EAF'] > 0.5,     # The IF condition because Minor Alleles need to be less than 0.5, if effect allele is less than 0.5, calculation does not apply and MAF = EAF
    1 - sa['EAF'],       # What to do if TRUE
    sa['EAF']            # What to do if FALSE
)
```

```python
sa
```

```python
# Find the lowest and highest EAF in your data, for sanity check if calculations were wrong
min_eaf = sa['EAF'].min() 
max_eaf = sa['EAF'].max()

print(f"The lowest EAF is: {min_eaf}")
print(f"The highest EAF is: {max_eaf}")
```

```python
sa = sa[sa['MAF'] >= 0.01] # Filtering SNPs with MAF less than 1%, theory is this may be only noise
```

```python
# Check for duplicate SNPs
print(sa.duplicated(subset=['CHR', 'BP']).sum())
```

```python
min_p = sa['P'].min() 
max_p = sa['P'].max()

print(f"The lowest p is: {min_p}")
print(f"The highest p is: {max_p}")
```

```python
sa = sa[(sa['P'] > 0) & (sa['P'] <= 1)]
```
