```python
# MAF calculation and cutoff using EAF number, >=1%, can be more lenient to the cutoff due to less reference for indonesian dataset
```

```python
trans['MAF'] = np.where(
    trans['EAF'] > 0.5,     # The IF condition because Minor Alleles need to be less than 0.5, if effect allele is less than 0.5, calculation does not apply and MAF = EAF
    1 - trans['EAF'],       # What to do if TRUE
    trans['EAF']            # What to do if FALSE
)
```

```python
trans
```

```python
trans = trans[trans['MAF'] >= 0.01] # Filtering SNPs with MAF less than 1%, theory is this may be only noise
```

```python
# Find the lowest and highest EAF in your data, for sanity check if calculations were wrong
min_eaf = trans['EAF'].min() 
max_eaf = trans['EAF'].max()

print(f"The lowest EAF is: {min_eaf}")
print(f"The highest EAF is: {max_eaf}")
```

```python
# Check for duplicate SNPs
print(trans.duplicated(subset=['CHR', 'BP']).sum())
```

```python
# P value cutoff where the P-value is greater than 0 and less than or equal to 1, can be more lenient to the cutoff due to less reference for indonesian dataset
trans = trans[(trans['P'] > 0) & (trans['P'] <= 1)]
```
