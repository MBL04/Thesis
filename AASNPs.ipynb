```python
import pandas as pd
import numpy as np
```

```python
path = r"C:\Users\micha\OneDrive\Documents\PRSGWAS_tsv\GCST90104550_buildGRCh37AA.tsv\Meta_IS_BothMandF_GIGA_AA_allfiles_metaAnalyseSEbased_wogc_2021June11_NeNs3rdDp1e6.gwascatalog.tsv_2"
gwas_AAdata = pd.read_csv(path, sep='\t')
display(gwas_AAdata)
```

```python
rename_dict = {
    'chromosome': 'CHR',
    'base_pair_location': 'BP',
    'effect_allele': 'A1',      # A1 is the target/risk allele
    'other_allele': 'A2',       # A2 is the baseline allele
    'beta': 'BETA',
    'p_value': 'P',
    'effect_allele_frequency': 'EAF',
    'standard_error': 'SE',
    'odds_ratio': 'OR'
}
```

```python
aarename = gwas_AAdata.rename(columns=rename_dict)
```

```python
aarename
```

```python
columns_to_keep = ['CHR', 'BP', 'A1', 'A2', 'BETA', 'P', 'EAF', 'SE', 'OR'] 
aa = aarename[columns_to_keep]
aa
```

```python
missing = aa.isna().sum()
print(missing)
```

```python
aa['MAF'] = np.where(
    aa['EAF'] > 0.5,     # The IF condition because Minor Alleles need to be less than 0.5, if effect allele is less than 0.5, calculation does not apply and MAF = EAF
    1 - aa['EAF'],       # What to do if TRUE
    aa['EAF']            # What to do if FALSE
)
```

```python
aa
```

```python
# Find the lowest and highest EAF in your data, for sanity check if calculations were wrong
min_eaf = aa['EAF'].min() 
max_eaf = aa['EAF'].max()

print(f"The lowest EAF is: {min_eaf}")
print(f"The highest EAF is: {max_eaf}")
```

```python
aa = aa[aa['MAF'] >= 0.01] # Filtering SNPs with MAF less than 1%, theory is this may be only noise
```

```python
# Check for duplicate SNPs
print(aa.duplicated(subset=['CHR', 'BP']).sum())
```

```python
min_p = aa['P'].min() 
max_p = aa['P'].max()

print(f"The lowest p is: {min_p}")
print(f"The highest p is: {max_p}")
```

```python
aa = aa[(aa['P'] > 0) & (aa['P'] <= 1)]
```
