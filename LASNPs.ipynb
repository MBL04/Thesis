```python
import pandas as pd
import numpy as np
```

```python
path = r"C:\Users\micha\OneDrive\Documents\PRSGWAS_tsv\GCST90104555_buildGRCh37LA.tsv\Meta_IS_BothMandF_GIGA_HA_allfiles_metaAnalyseSEbased_wogc_2021June11_NeNs3rdDp1e6.gwascatalog.tsv_2"
gwas_LAdata = pd.read_csv(path, sep='\t')
display(gwas_LAdata)
```

```python
rename_dict = {
    'chromosome': 'CHR',
    'base_pair_location': 'BP',
    'effect_allele': 'A1',      # A1 is the target/risk allele
    'other_allele': 'A2',       # A2 is the baseline allele
    'beta': 'BETA',
    'p_value': 'P',
    'effect_allele_frequency': 'EAF',
    'standard_error': 'SE',
    'odds_ratio': 'OR'
}
```

```python
larename = gwas_LAdata.rename(columns=rename_dict)
```

```python
larename
```

```python
columns_to_keep = ['CHR', 'BP', 'A1', 'A2', 'BETA', 'P', 'EAF', 'SE', 'OR'] 
la = larename[columns_to_keep]
la
```

```python
missing = la.isna().sum()
print(missing)
```

```python
la['MAF'] = np.where(
    la['EAF'] > 0.5,     # The IF condition because Minor Alleles need to be less than 0.5, if effect allele is less than 0.5, calculation does not apply and MAF = EAF
    1 - la['EAF'],       # What to do if TRUE
    la['EAF']            # What to do if FALSE
)
```

```python
la
```

```python
# Find the lowest and highest EAF in your data, for sanity check if calculations were wrong
min_eaf = la['EAF'].min() 
max_eaf = la['EAF'].max()

print(f"The lowest EAF is: {min_eaf}")
print(f"The highest EAF is: {max_eaf}")
```

```python
la = la[la['MAF'] >= 0.01] # Filtering SNPs with MAF less than 1%, theory is this may be only noise
```

```python
# Check for duplicate SNPs
print(la.duplicated(subset=['CHR', 'BP']).sum())
```

```python
min_p = la['P'].min() 
max_p = la['P'].max()

print(f"The lowest p is: {min_p}")
print(f"The highest p is: {max_p}")
```

```python
la = la[(la['P'] > 0) & (la['P'] <= 1)]
```
